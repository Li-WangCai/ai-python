### 1. 用全连接神经网络判断数字

```python
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
import matplotlib.pylab as plt
%matplotlib inline
```

```python
# 检查 CUDA 是否可用
if torch.cuda.is_available():
    print(f"GPU 可用，设备数量: {torch.cuda.device_count()}")
    print(f"当前设备: {torch.cuda.current_device()}")
    print(f"设备名称: {torch.cuda.get_device_name(0)}")
else:
    print("GPU 不可用，将使用 CPU")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
input_size = 784
hidden_size = 100
num_classes = 10
num_epochs = 5
batch_size = 100
learning_rate = 0.001
```

```python
# 加载MNIST数据 
train_dataset = torchvision.datasets.MNIST(root='../data/MNIST_data', #数据存在这里
                                           train=True, #标记为训练数据
                                           transform=transforms.ToTensor(),#转换成torch.tensor
                                           download=True)#不够就下载

test_dataset = torchvision.datasets.MNIST(root='../data/MNIST_data', 
                                          train=False, #标记成测试数据
                                          transform=transforms.ToTensor())

```

```python
train_loader = torch.utils.data.DataLoader(dataset=train_dataset, #去这里读数据
                                           batch_size=batch_size, #按照每批几个去读
                                           shuffle=True) #通过洗牌随机化batch

test_loader = torch.utils.data.DataLoader(dataset=test_dataset, #和上面一样，只是这里搞测试数据
                                          batch_size=batch_size, 
                                          shuffle=False)
```

```python
for i,(images,labels) in  enumerate(train_loader):
    plt.imshow(images[0,0,:,:],cmap='gray');#显示shuffle过后第一幅图，所以是随机的
    if i > 1:
        print(images.shape)#[100,1,28,28]100个图，每个图下面只有一个子图，每个子图是一个28*28的数组
        print(labels.shape)#[100] 100个label
        break
#print(labels)#每个图形对应的数字
```

```python

class NeuralNet(nn.Module):
    def __init__(self, input_size, hidden_size, num_classes):#input_size图像尺寸，hidden_size隐藏层大小，每个隐藏层都一样大，num_classes最后分类数（10）
        super(NeuralNet, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size) #输入到第一隐藏层
        self.relu = nn.ReLU() #激活函数
        self.fc2 = nn.Linear(hidden_size, num_classes)#第一隐藏层到第二隐藏层  
    
    def forward(self, x):
        out = self.fc1(x) #输入到第一
        out = self.relu(out)#第一到第二
        out = self.fc2(out)#第二到输出分类
        return out
```

```python
model = NeuralNet(input_size, hidden_size, num_classes)#一个神经网络实例
model = model.to(device) #模型转移到gpu上算
```

```python
# Loss and optimizer
criterion = nn.CrossEntropyLoss()#损失函数
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)  #优化方法
```

```python

total_step = len(train_loader)
for epoch in range(num_epochs):
    for i, (images, labels) in enumerate(train_loader):  
        images = images.reshape(-1, 28*28)      #-1表示数据长度，也就是最后一个元素的位置
        images=images.to(device) #数据上gpu
        labels=labels.to(device)  #标签上gpu
        outputs = model(images)
        loss = criterion(outputs, labels)
        
        optimizer.zero_grad()#梯度清零
        loss.backward()#后向演化计算梯度
        optimizer.step()#用梯度走一步
        
        if (i+1) % 100 == 0:
            print ('Epoch [{}/{}], Step [{}/{}], Loss: {:.4f}' 
                   .format(epoch+1, num_epochs, i+1, total_step, loss.item()))
```

```python
with torch.no_grad():#关掉梯度计算功能可能快一点
    correct = 0
    total = 0
    for images, labels in test_loader:
        images = images.reshape(-1, 28*28)
        images=images.to(device)
        labels=labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs, 1) #outputs 每一行是一个概率数组，表示图像等于1的概率2的概率等等，取其中最大一个最为我们的预测，每列都这样做
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    print('Accuracy of the network on the 10000 test images: {} %'.format(100 * correct / total))
```

### 3. 用卷积神经网络判断服饰

```python

num_epochs = 5
num_classes = 10
batch_size = 100
learning_rate = 0.001
```

```python

class ConvNet(nn.Module):
    def __init__(self, num_classes=10):#定义类的初始化方法，接受一个参数num_classes（类别数），默认值为10（对应MNIST/Fashion-MNIST的10个类别）。
        super(ConvNet, self).__init__()#初始化卷积神经网络父类
        self.cnn_layer = nn.Sequential( #按顺序执行2d卷积层，ReLU激活，最大化池化
            nn.Conv2d(1, 32, kernel_size=5, stride=1, padding=2),#卷积层
            nn.ReLU(),#激活函数
            nn.MaxPool2d(kernel_size=2, stride=2)) #池化
        self.fc = nn.Linear(14*14*32, num_classes) #添加全连接层做最后全局判断，一共10个分类，输出每个类的概率
        
    def forward(self, x):
        out = self.cnn_layer(x) #原始数据进来经过cnn_layer
        out = out.reshape(out.size(0), -1)#out.size(0)是一个batch的数量，然后自动计算每个图拍扁后一维向量是多大
        out = self.fc(out)#矢量化计算batch里面每个图的概率分布
        return out
'''
nn.Conv2d(1, 32, kernel_size=5, stride=1, padding=2),
添加2D卷积层：
1：输入通道数（灰度图像，单通道）
32：输出通道数（使用32个不同的卷积核）
kernel_size=5：卷积核大小为5×5
stride=1：步长为1
padding=2：边缘填充2行/列（输入为28×28时，输出保持28×28）
输入尺寸计算：对于28×28输入，输出为(28-5+2×2)/1 + 1 = 28×28

nn.MaxPool2d(kernel_size=2, stride=2))
添加2D最大池化层：
kernel_size=2：池化窗口大小为2×2
stride=2：步长为2（窗口不重叠）,步长是窗口每次移动的pixels个数
尺寸计算：28×28输入 → 池化后变为14×14

self.fc = nn.Linear(14 * 14 * 32, num_classes)
添加全连接层（分类器）：
14 * 14 * 32：输入特征维度（14×14×32 = 6272）
num_classes：输出维度（对应10个类别的得分）

out = out.reshape(out.size(0), -1)
每个28*28的图，经过32个卷积核，得到32个28*28的图，每个图池化变成14*14的图，最终得到32个14*14的特征图
将特征图展平为一维向量：
out.size(0)：批量大小（batch_size），一定记住为了泛化，参数不是对单独一个图优化的，而是对一个batch优化的
-1：自动计算展平后的维度（14×14×32 = 6272）
例如，形状从[batch_size, 32, 14, 14]变为[batch_size, 6272]


'''
```

```python
model_fashion_cnn = ConvNet(num_classes)
criterion = nn.CrossEntropyLoss() #定义损失函数为交叉熵损失，常用于多分类问题。它会将模型输出（logits）转换为概率分布，再计算损失。
optimizer = torch.optim.Adam(model_fashion_cnn.parameters(), lr=learning_rate)
```

```python
total_step = len(train_loader)
for epoch in range(num_epochs):
    for i, (images, labels) in enumerate(train_loader):
        
        outputs = model_fashion_cnn(images)#由forward计算的到的一个batch里面每个图的分类概率矢量
        loss = criterion(outputs, labels)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        if (i+1) % 100 == 0:
            print ('Epoch [{}/{}], Step [{}/{}], Loss: {:.4f}' 
                   .format(epoch+1, num_epochs, i+1, total_step, loss.item()))
```

```python
with torch.no_grad():
    correct = 0
    total = 0
    for images, labels in test_loader:
        outputs = model_fashion_cnn(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    print('Test Accuracy of the model on the 10000 test images: {} %'.format(100 * correct / total))
```
